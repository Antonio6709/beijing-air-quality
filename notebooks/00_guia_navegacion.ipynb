## 🚀 Pipeline de Ejecución Maestra
La siguiente celda permite ejecutar todos los notebooks en orden secuencial. Esto asegurará que:
1. Se procesen los datos crudos y se guarden limpios en la carpeta correcta.
2. Se actualicen las visualizaciones y las tablas SQL.

```python
import nbformat
from nbconvert.preprocessors import ExecutePreprocessor
import os
import time

def ejecutar_pipeline_completo():
    # Lista de notebooks adaptada a la nueva estructura de 3 archivos
    notebooks = [
        "01_limpieza_y_unificacion.ipynb",
        "02_eda_y_visualizacion.ipynb",
        "03_sql_y_resumen_ejecutivo.ipynb"
    ]
    
    print("🔥 Iniciando ejecución maestra del proyecto...")
    inicio_total = time.time()

    for nb_name in notebooks:
        if not os.path.exists(nb_name):
            print(f"⚠️ Aviso: No se encontró {nb_name}. Saltando...")
            continue
            
        print(f"⏳ Ejecutando {nb_name}...", end=" ")
        inicio_nb = time.time()
        
        try:
            with open(nb_name, 'r', encoding='utf-8') as f:
                nb = nbformat.read(f, as_version=4)
            
            # Ejecutor de celdas
            ep = ExecutePreprocessor(timeout=600, kernel_name='python3')
            ep.preprocess(nb, {'metadata': {'path': './'}})
            
            # Guardamos para mantener los outputs actualizados
            with open(nb_name, 'w', encoding='utf-8') as f:
                nbformat.write(nb, f)
                
            print(f"✅ OK ({time.time() - inicio_nb:.1f}s)")
            
        except Exception as e:
            print(f"❌ Error en {nb_name}: {e}")
            break

    print(f"\n✨ Finalizado en {time.time() - inicio_total:.1f} segundos.")
    print("📈 El pipeline se ha completado exitosamente con la nueva arquitectura modular.")

# Ejecutar el botón maestro
ejecutar_pipeline_completo()
```
