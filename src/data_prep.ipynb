```python
import pandas as pd
import glob
import os

def load_and_unify_data(data_path):
    """
    Lee todos los CSVs de la ruta especificada y los concatena en un solo DataFrame.
    """
    all_files = glob.glob(os.path.join(data_path, "*.csv"))
    df_list = [pd.read_csv(file) for file in all_files]
    
    df_unified = pd.concat(df_list, ignore_index=True)
    return df_unified
```
