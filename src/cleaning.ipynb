```python
import pandas as pd

def clean_and_interpolate(df):
    """
    Ordena el dataset, crea el índice temporal y aplica interpolación lineal 
    para rellenar los valores nulos (Na/NaN) agrupando por estación.
    """
    # Crear variable datetime a partir de las columnas de fecha y hora
    df['date'] = pd.to_datetime(df[['year', 'month', 'day', 'hour']])
    df = df.sort_values(by=['station', 'date'])
    df = df.set_index('date')
    
    # Interpolación lineal por estación para los nulos
    df = df.groupby('station').apply(lambda group: group.interpolate(method='linear'))
    
    # Limpiar columnas innecesarias si existen
    cols_to_drop = ['year', 'month', 'day', 'hour', 'No']
    df = df.drop(columns=[c for c in cols_to_drop if c in df.columns], errors='ignore')
    
    # Eliminar posibles nulos residuales al principio de las series
    df = df.dropna()
    
    return df
```
